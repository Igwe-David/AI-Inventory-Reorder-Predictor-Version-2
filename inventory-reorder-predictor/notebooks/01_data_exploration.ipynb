Run from the project folder: `pip install notebook` then `jupyter notebook`. Run the cells top to bottom.

```python
import sys, os
sys.path.insert(0, os.path.abspath('..'))
import pandas as pd
import matplotlib.pyplot as plt
from src import config
raw = pd.read_csv(config.RAW_FILE, parse_dates=['date'])
raw.head()
```

## Size and missing values

```python
print(raw.shape)
raw.isna().sum()
```

```python
raw.describe().T
```

## Average daily sales by category

```python
raw.groupby('category')['units_sold'].mean().sort_values().plot.barh(title='Average units sold per day')
plt.show()
```

## Weekly pattern (0 = Monday ... 6 = Sunday)

```python
raw.groupby('day_of_week')['units_sold'].mean().plot.bar(title='Average units sold by day of week')
plt.show()
```

## Monthly pattern (note the December rush)

```python
raw.groupby('month')['units_sold'].mean().plot.bar(title='Average units sold by month')
plt.show()
```

## One product over time

```python
p = raw[raw['product_name'] == 'Bottled water (pack)'].set_index('date')
p['units_sold'].rolling(7).mean().plot(figsize=(10,3), title='Bottled water - 7-day average sales')
plt.show()
```

## Stock-outs: days when customers could not be served
Sales are cut off on these days, so recorded sales UNDERSTATE true demand.

```python
print(f"Stock-out days: {raw['stockout_flag'].mean():.1%}")
raw.groupby('product_name')['stockout_flag'].mean().sort_values().plot.barh(figsize=(6,6))
plt.show()
```
